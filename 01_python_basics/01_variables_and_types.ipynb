## 1.1 Variables

**What is a variable?**
A variable is simply a name that points to a value stored in memory. Think of it like a labelled box — the label is the name, and whatever is inside is the value.

Python is **dynamically typed**, meaning you do not need to declare what *type* of value a variable holds. Python figures that out automatically when you assign a value.

**Naming rules:** Variable names must start with a letter or underscore, contain only letters/numbers/underscores, and cannot be Python keywords like `if`, `for`, `class`, etc.

**Best practice:** Use descriptive `snake_case` names — `student_age` not `x`, `sa`, or `StudentAge`.

```python
# Assigning variables — Python infers the type automatically
name = "Alice"          # str (text in quotes)
age = 25                # int (whole number)
height = 5.6            # float (decimal number)
is_student = True       # bool (True or False)
middle_name = None      # NoneType (represents "no value")

print(name, age, height, is_student, middle_name)
print(type(name), type(age), type(height), type(is_student), type(middle_name))

# Multiple assignment on one line
x, y, z = 1, 2, 3
print(x, y, z)

# Swap two values — Python makes this elegant
a, b = 10, 20
a, b = b, a
print(f"After swap: a={a}, b={b}")
```

## 1.2 Built-in Data Types

**The five core types:**

| Type | Example | What it represents |
|------|---------|-------------------|
| `int` | `42`, `-7`, `0` | Whole numbers, no limit to size in Python |
| `float` | `3.14`, `-0.5`, `2.0` | Decimal numbers (64-bit floating point) |
| `str` | `"hello"`, `'world'` | Text, enclosed in single or double quotes |
| `bool` | `True`, `False` | Logical values (note: capital T and F) |
| `NoneType` | `None` | Represents the absence of a value |

**Type conversion (casting):** You often need to convert between types — e.g. input from a user always arrives as a string, but you need a number to do maths.

```python
# Checking types
print(type(42))          # <class 'int'>
print(type(3.14))        # <class 'float'>
print(type("hello"))     # <class 'str'>
print(type(True))        # <class 'bool'>
print(type(None))        # <class 'NoneType'>

# Type conversion — converting one type to another
x = "42"                 # a string, not a number yet
print(type(x), x + "8") # string concatenation: "428" (NOT maths!)

x_as_int = int(x)
print(type(x_as_int), x_as_int + 8)  # now it's maths: 50

# More conversions
print(float("3.14"))    # string -> float
print(str(100))         # int -> string
print(int(9.9))         # float -> int: truncates (does NOT round)
print(int(9.1))         # also 9 — truncate means "chop off decimal"
print(bool(0))          # 0 is False
print(bool(42))         # any non-zero is True
print(bool(""))         # empty string is False
print(bool("hello"))    # non-empty string is True
```

## 1.3 Strings In Depth

**What is a string?**
A string is a sequence of characters. Strings are **immutable** — you cannot change individual characters in a string; you create a new string instead.

**Indexing:** Each character has a position (index) starting from 0. Negative indices count backwards from the end (-1 is the last character).

**Slicing:** Extract a portion of a string using `string[start:stop:step]`. The character at `stop` index is NOT included. If start/stop are omitted, they default to the beginning/end.

```python
s = "Hello, World!"

# Indexing — accessing individual characters
print(s[0])       # 'H'  — first character (index 0)
print(s[7])       # 'W'
print(s[-1])      # '!'  — last character (-1 = one from the end)
print(s[-6])      # 'W'  — 6th from the end

# Slicing — s[start:stop] extracts characters from start up to (not including) stop
print(s[0:5])     # 'Hello'   — characters 0,1,2,3,4
print(s[7:12])    # 'World'
print(s[:5])      # 'Hello'   — omitting start defaults to 0
print(s[7:])      # 'World!'  — omitting stop goes to the end
print(s[::2])     # every 2nd character
print(s[::-1])    # reversed string — step of -1 goes backwards
```

```python
# String methods — operations you can perform on strings
greeting = "  hello, world!  "

print(greeting.strip())            # remove leading/trailing whitespace
print(greeting.strip().title())    # title case: 'Hello, World!'
print(greeting.strip().upper())    # 'HELLO, WORLD!'
print(greeting.strip().lower())    # 'hello, world!'
print(greeting.strip().replace(",", " and"))   # replace substring
print("hello world".split())      # split on spaces -> ['hello', 'world']
print(",".join(["a", "b", "c"]))  # join with separator -> 'a,b,c'
print("hello".startswith("he"))   # True
print("hello".endswith("lo"))     # True
print(len("hello world"))         # 11 — length including space
```

## 1.4 f-Strings — Modern String Formatting

**What are f-strings?**
f-strings (formatted string literals) let you embed Python expressions directly inside a string by placing them inside `{}`. They are the modern, recommended way to format strings — more readable and faster than older alternatives like `%` formatting or `.format()`.

**Format specifiers:** After the variable name, add `:` followed by a format spec to control how the value is displayed. Common ones: `.2f` = 2 decimal places, `d` = integer, `,` = thousands separator, `<` `>` `^` = alignment.

```python
name = "Alice"
score = 95.6789
rank = 1
students_total = 42

# Basic embedding
print(f"Student: {name}")

# Expressions inside f-strings
print(f"Score: {score:.2f}")             # 2 decimal places
print(f"Score: {score:.0f}")             # rounded to integer
print(f"Rank: {rank} of {students_total}")
print(f"Top {rank/students_total*100:.1f}% of the class")

# Padding and alignment
for item, price in [("Apple", 1.5), ("Banana", 0.75), ("Cherry", 3.0)]:
    print(f"  {item:<10} £{price:>6.2f}")   # left-align item, right-align price

# Multiline f-string (use triple quotes)
report = f"""
Student Report
==============
Name:  {name}
Score: {score:.1f}/100
Rank:  {rank}/{students_total}
"""
print(report)
```

## 1.5 Arithmetic Operators

**Python's arithmetic operators:**

| Operator | Example | Result | Notes |
|----------|---------|--------|-------|
| `+` | `7 + 3` | `10` | Addition |
| `-` | `7 - 3` | `4` | Subtraction |
| `*` | `7 * 3` | `21` | Multiplication |
| `/` | `7 / 3` | `2.333...` | True division — **always returns float** |
| `//` | `7 // 3` | `2` | Floor division — rounds DOWN to whole number |
| `%` | `7 % 3` | `1` | Modulo — the **remainder** after division |
| `**` | `7 ** 3` | `343` | Exponentiation (7³) |

**Key distinction:** `/` always gives a float. `//` gives an int (or float if either operand is float).

```python
a, b = 10, 3

print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")    # 3.333...  (always float!)
print(f"{a} // {b} = {a // b}")  # 3         (floor division)
print(f"{a} % {b} = {a % b}")    # 1         (remainder)
print(f"{a} ** {b} = {a ** b}")  # 1000

# Real uses:
print()
print("Practical uses of operators:")
print(f"  Is 42 even? {42 % 2 == 0}")              # modulo to check even/odd
print(f"  Minutes in 127 mins: {127 // 60}h {127 % 60}m")  # floor div + modulo
print(f"  2^10 = {2 ** 10}")                       # powers of 2

# Augmented assignment operators (shortcuts)
x = 10
x += 5    # same as x = x + 5
print(f"x = {x}")
x *= 2    # same as x = x * 2
print(f"x = {x}")
```

## 1.6 Comparison and Logical Operators

**Comparison operators** compare two values and always return `True` or `False`.

**Logical operators** combine multiple conditions:
- `and` — both conditions must be True
- `or` — at least one condition must be True
- `not` — inverts True/False

**Python's chained comparisons** are a unique and readable feature: `10 < x < 20` is valid Python (and equivalent to `x > 10 and x < 20`).

**Truthiness:** In Python, every value has a truth value. `False`, `0`, `0.0`, `""`, `[]`, `{}`, `None` are all "falsy". Everything else is "truthy". This is used constantly in conditions.

```python
x = 15

# Comparison operators
print(x > 10)      # True
print(x < 10)      # False
print(x == 15)     # True   (equality, note: == not =)
print(x != 10)     # True   (not equal)
print(x >= 15)     # True   (greater than or equal)
print(x <= 14)     # False

# Logical operators
print(x > 10 and x < 20)   # True — both are True
print(x < 5 or x == 15)    # True — at least one is True
print(not x == 15)          # False — not inverts True to False

# Python's elegant chained comparison
print(10 < x < 20)          # True — reads like a maths inequality!

# Truthiness examples
print(bool(0), bool(1))         # False True
print(bool(""), bool("hi"))     # False True
print(bool([]), bool([1,2]))    # False True
print(bool(None), bool(3.14))   # False True
```

## 1.7 Getting User Input

**`input()` function:** Pauses the program and waits for the user to type something. It ALWAYS returns a string — even if the user types a number. You must convert it yourself.

**In notebooks:** `input()` works, but for demos we assign values directly to keep examples reproducible.

```python
# Simulating user input (in a real script, use: name = input("Enter your name: "))
name = "Bob"
age_str = "22"   # input() always returns a string

# Convert to int for arithmetic
age = int(age_str)

print(f"Hello {name}! You are {age} years old.")
print(f"You were born around {2024 - age}.")
print(f"In 10 years you will be {age + 10}.")

# Common mistake to avoid:
# age_str + 10  would raise TypeError — can't add str and int
# int(age_str) + 10  works correctly
```
