### Exercise 1 — Variables and f-strings

**Concept:** Variables store values. f-strings let you embed variable values directly inside a string using `{variable_name}` syntax.

```python
name = "Alice"
age = 25
favourite_language = "Python"

print(f"My name is {name}, I am {age} years old, and my favourite language is {favourite_language}.")

# You can also do calculations inside f-strings:
print(f"In 5 years I will be {age + 5} years old.")
```

**Why this works:** The `f` before the quote tells Python this is an f-string. Anything inside `{}` is treated as Python code and its result is inserted into the string.

### Exercise 2 — Type conversion

**Concept:** Python cannot do arithmetic on strings, even if they look like numbers. You must explicitly convert them using `int()`, `float()`, etc.

```python
pi_string = '3.14159'

# Step 1: convert string to float
pi_float = float(pi_string)
print(type(pi_float))   # confirms it's now a float

# Step 2: multiply
result = pi_float * 2

# Step 3: format to 3 decimal places using :.3f inside f-string
print(f"Result: {result:.3f}")

# One-liner version:
print(f"{float('3.14159') * 2:.3f}")
```

**Why `:.3f`?** Inside an f-string, after the variable name you can add `:` followed by a format spec. `f` means fixed-point (decimal), `3` means 3 digits after the decimal point.

### Exercise 3 — String slicing

**Concept:** Strings are indexed from 0. Slicing syntax is `string[start:stop]`. The character at `stop` index is NOT included.

```python
s = 'Data Science is awesome!'

# Method 1: Count the characters
# D=0, a=1, t=2, a=3, (space)=4, S=5, c=6, i=7, e=8, n=9, c=10, e=11
# 'Science' starts at index 5 and ends at index 12
print(s[5:12])

# Method 2: Use .find() to locate the start position automatically (more robust)
start = s.find('Science')
end = start + len('Science')
print(s[start:end])

# Verify
print(s.find('Science'))   # shows index 5
```

**Key insight:** Always verify your slice indices by counting from 0. Using `.find()` is safer when the position might change.

### Exercise 4 — Chained comparison

**Concept:** Python supports chaining comparisons like `10 < n < 20`, which reads naturally and is equivalent to `n > 10 and n < 20`.

```python
n = 17

# Pythonic chained comparison (preferred)
result = 10 <= n <= 20
print(result)   # True

# Equivalent using 'and'
result2 = n >= 10 and n <= 20
print(result2)  # True — same thing

# Test with a value outside range
m = 25
print(10 <= m <= 20)   # False
```

### Exercise 5 — Arithmetic and f-string formatting

**Concept:** Use variables to store intermediate values, then format the output string precisely.

```python
price_per_apple = 0.50
quantity = 13

total = price_per_apple * quantity
print(f'{quantity} apples cost £{total:.2f}')

# Why :.2f? Currency always needs exactly 2 decimal places.
# Without it: £6.5  — looks wrong
# With :.2f:  £6.50 — correct
```

### Exercise 6 — String method chaining

**Concept:** String methods can be chained because each one returns a new string. `.strip()` removes whitespace. `.lower()` converts to lowercase. `.replace()` swaps characters.

```python
text = '  HELLO WORLD  '

# Step 1: strip whitespace, Step 2: lowercase
step1 = text.strip().lower()
print(repr(step1))   # 'hello world'

# Step 2: replace spaces with underscores
step2 = step1.replace(' ', '_')
print(step2)   # 'hello_world'

# One-line chained version:
result = '  HELLO WORLD  '.strip().lower().replace(' ', '_')
print(result)

# Why repr()? It shows the string with quotes, so you can see spaces clearly.
```

### Exercise 7 — Predicting outputs

**Concept:** Before running code, predicting output is a powerful learning technique. It forces you to understand operator behaviour.

```python
x = 7
print(x // 2)   # 3  — floor division: 7/2 = 3.5, round DOWN to 3
print(x % 2)    # 1  — remainder: 7 = 3*2 + 1, so remainder is 1
print(x ** 2)   # 49 — exponentiation: 7 * 7
print(x / 2)    # 3.5 — true division, always returns float

# bool() truthiness:
# False values: 0, empty string "", empty list [], empty dict {}, None
# True values: anything else
print(bool(0), bool(1), bool(""), bool("hi"))   # False True False True
```

### Exercise 8 — FizzBuzz

**Concept:** This classic problem uses the modulo operator `%` to check divisibility. Check the most specific condition first (divisible by BOTH), then the individual ones.

```python
for n in range(1, 16):
    if n % 3 == 0 and n % 5 == 0:   # must check BOTH first
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)

# Common mistake: checking 3 and 5 separately first means
# you never reach the FizzBuzz condition for 15.
```

**Common pitfall:** If you check `n % 3 == 0` first and print 'Fizz', Python skips the rest — so 15 would print 'Fizz' instead of 'FizzBuzz'. Always check the combined condition first.
