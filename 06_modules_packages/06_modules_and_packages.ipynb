## 6.1 Importing Modules

````python
# Standard library modules (no install needed)
import math
import random
import datetime
import os

print(math.pi)
print(math.sqrt(144))
print(random.randint(1, 100))
print(datetime.date.today())
````

````python
# Import specific names
from math import pi, sqrt, factorial
from random import choice, shuffle

print(pi)
print(sqrt(16))
print(factorial(5))

colours = ["red", "green", "blue"]
print(choice(colours))
````

````python
# Import with alias (very common in data science!)
import numpy as np          # convention: np
import pandas as pd         # convention: pd
import matplotlib.pyplot as plt  # convention: plt

# Check if available
for lib in ["numpy", "pandas", "matplotlib"]:
    try:
        __import__(lib)
        print(f"✅ {lib} available")
    except ImportError:
        print(f"❌ {lib} — install with: pip install {lib}")
````

## 6.3 Virtual Environments
```bash
# Create a virtual environment
python -m venv venv

# Activate it
source venv/bin/activate        # macOS / Linux
venv\\Scripts\\activate          # Windows

# Install packages
pip install numpy pandas scikit-learn

# Save your dependencies
pip freeze > requirements.txt

# Install from requirements file
pip install -r requirements.txt

# Deactivate
deactivate
```

> **Why virtual environments?** They isolate your project's dependencies from other projects and from the system Python, preventing version conflicts.
