# 12 — Data Visualisation
**Learning Objectives:** Matplotlib basics, Seaborn statistical plots, choosing the right chart
---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Use a clean style
plt.style.use("seaborn-v0_8")
sns.set_palette("husl")

# Sample data for all examples
np.random.seed(42)
df = pd.DataFrame({
    "x": np.linspace(0, 10, 100),
    "y": np.sin(np.linspace(0, 10, 100)) + np.random.normal(0, 0.1, 100),
    "category": np.random.choice(["A", "B", "C"], 100),
    "value": np.random.normal(50, 15, 100)
})
print(df.head())
```

## 12.1 Line & Scatter Plots

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Line plot
axes[0].plot(df["x"], df["y"], color="steelblue", linewidth=2)
axes[0].set_title("Line Plot: sin(x) + noise")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")
axes[0].axhline(0, color="gray", linestyle="--", alpha=0.5)

# Scatter plot
axes[1].scatter(df["x"], df["y"], c=df["category"].astype("category").cat.codes,
                cmap="viridis", alpha=0.7, s=30)
axes[1].set_title("Scatter Plot by Category")
axes[1].set_xlabel("x")
axes[1].set_ylabel("y")

plt.tight_layout()
plt.savefig("/tmp/line_scatter.png", dpi=100, bbox_inches="tight")
plt.show()
print("Plot saved!")
```

## 12.2 Distribution Plots

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Histogram
axes[0].hist(df["value"], bins=20, color="steelblue", edgecolor="white")
axes[0].set_title("Histogram")
axes[0].set_xlabel("value")

# KDE (density curve)
sns.kdeplot(df["value"], ax=axes[1], fill=True, color="coral")
axes[1].set_title("KDE Plot")

# Boxplot by category
sns.boxplot(data=df, x="category", y="value", ax=axes[2], palette="Set2")
axes[2].set_title("Boxplot by Category")

plt.tight_layout()
plt.savefig("/tmp/distributions.png", dpi=100, bbox_inches="tight")
plt.show()
```

## 12.3 Categorical Plots

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Bar chart
cat_means = df.groupby("category")["value"].mean()
axes[0].bar(cat_means.index, cat_means.values, color=["#e74c3c","#3498db","#2ecc71"])
axes[0].set_title("Mean Value by Category")
axes[0].set_ylabel("Mean Value")

# Violin plot
sns.violinplot(data=df, x="category", y="value", ax=axes[1], palette="muted")
axes[1].set_title("Distribution by Category (Violin)")

plt.tight_layout()
plt.savefig("/tmp/categorical.png", dpi=100, bbox_inches="tight")
plt.show()
```

## 12.4 Heatmaps & Correlation (Key for ML EDA!)

```python
# Generate correlated data
np.random.seed(42)
n = 200
ml_df = pd.DataFrame({
    "age":     np.random.normal(35, 10, n),
    "income":  np.random.normal(50000, 15000, n),
    "score":   np.random.normal(70, 10, n),
    "tenure":  np.random.normal(5, 3, n).clip(0),
})
ml_df["performance"] = 0.3*ml_df["score"] + 0.2*ml_df["tenure"] + np.random.normal(0,5,n)

# Correlation heatmap
fig, ax = plt.subplots(figsize=(7, 5))
corr = ml_df.corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm",
            center=0, square=True, ax=ax)
ax.set_title("Feature Correlation Heatmap")
plt.tight_layout()
plt.savefig("/tmp/heatmap.png", dpi=100, bbox_inches="tight")
plt.show()
```

## 12.5 Choosing the Right Chart

| Data | Chart |
|------|-------|
| Continuous over time/order | Line chart |
| Two continuous variables | Scatter plot |
| Distribution of one variable | Histogram or KDE |
| Comparing categories | Bar chart |
| Distribution by category | Box or Violin plot |
| Relationships between many variables | Heatmap / Pair plot |
| Part-to-whole | Pie chart *(use sparingly)* |
