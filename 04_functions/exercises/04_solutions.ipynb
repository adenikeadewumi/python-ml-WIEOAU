### Exercise 1 — Celsius to Fahrenheit function

**Concept:** A function takes input, does something, and returns output. The `return` statement sends the result back to the caller. Docstrings document what the function does.

```python
def celsius_to_fahrenheit(c):
    """Convert Celsius temperature to Fahrenheit.
    
    Args:
        c: Temperature in Celsius (int or float)
    Returns:
        Temperature in Fahrenheit (float)
    """
    return c * 9/5 + 32

# Test it
for temp_c in [0, 100, 37]:
    temp_f = celsius_to_fahrenheit(temp_c)
    print(f"{temp_c}°C = {temp_f:.1f}°F")

# Extra: make it work both ways
def convert_temp(value, unit="C"):
    """Convert temperature. unit='C' converts to F, unit='F' converts to C."""
    if unit.upper() == "C":
        return value * 9/5 + 32
    else:
        return (value - 32) * 5/9

print(convert_temp(100, "C"))   # 212.0
print(convert_temp(212, "F"))   # 100.0
```

### Exercise 2 — Palindrome checker

**Concept:** Strip spaces, convert to lowercase, then compare the string to its reverse. In Python, `string[::-1]` reverses a string — this is slicing with step -1.

```python
def is_palindrome(s):
    """Return True if s reads the same forwards and backwards (ignores case and spaces)."""
    # Clean: lowercase and remove spaces
    cleaned = s.lower().replace(" ", "")
    # Compare to its reverse
    return cleaned == cleaned[::-1]

# Test cases
tests = [
    ("racecar", True),
    ("A man a plan a canal Panama", True),
    ("hello", False),
    ("Was it a car or a cat I saw", True),
]

for text, expected in tests:
    result = is_palindrome(text)
    status = "PASS" if result == expected else "FAIL"
    print(f"[{status}] '{text[:30]}' -> {result}")
```

**`[::-1]` explained:** This is slice notation `[start:stop:step]`. Leaving start and stop blank means 'the whole string'. Step of `-1` means go backwards, which reverses the string.

### Exercise 3 — Count vowels

**Concept:** Use a set for O(1) membership testing (faster than a list). Loop through characters and count those that are vowels.

```python
def count_vowels(text):
    """Return the number of vowels (a, e, i, o, u) in text (case-insensitive)."""
    vowels = set("aeiou")   # set lookup is O(1) — fast!
    count = 0
    for char in text.lower():
        if char in vowels:
            count += 1
    return count

    # One-liner version:
    # return sum(1 for c in text.lower() if c in set("aeiou"))

# Tests
print(count_vowels("Hello World"))           # 3
print(count_vowels("Python is awesome"))     # 7
print(count_vowels("rhythm"))                # 0
```

### Exercise 4 — HTML tag builder with **kwargs

**Concept:** `**kwargs` collects all keyword arguments into a dictionary. You can iterate over them to build dynamic strings. The trailing underscore in `class_` is a convention to avoid clashing with Python's built-in `class` keyword.

```python
def make_tag(tag, **attributes):
    """Build an HTML opening tag with given attributes."""
    if not attributes:
        return f"<{tag}>"
    
    # Build attribute string
    attr_parts = []
    for key, value in attributes.items():
        # Remove trailing underscore (e.g. class_ -> class)
        clean_key = key.rstrip("_")
        attr_parts.append(f"{clean_key}='{value}'")
    
    attr_string = " ".join(attr_parts)
    return f"<{tag} {attr_string}>"

# Tests
print(make_tag("a", href="https://python.org", class_="link"))
print(make_tag("input", type="text", id="username", placeholder="Enter name"))
print(make_tag("img", src="photo.jpg", alt="A photo", width="300"))
print(make_tag("br"))   # no attributes
```

### Exercise 5 — Challenge: Memoize decorator

**Concept:** A decorator wraps a function to add behaviour. A memoization cache stores previously computed results — when the same input is seen again, return the cached result instead of recomputing.

```python
import time

def memoize(func):
    """Cache the results of a function call. Returns cached result on repeat inputs."""
    cache = {}   # dict to store {args: result}
    
    def wrapper(*args):
        if args not in cache:
            cache[args] = func(*args)   # compute and store
        return cache[args]              # return from cache
    
    wrapper.__name__ = func.__name__    # preserve function name
    return wrapper

# Without memoization — SLOW (recalculates same values over and over)
def fib_slow(n):
    if n <= 1:
        return n
    return fib_slow(n-1) + fib_slow(n-2)

# With memoization — FAST
@memoize
def fib_fast(n):
    if n <= 1:
        return n
    return fib_fast(n-1) + fib_fast(n-2)

# Compare speed
n = 35

start = time.time()
print(f"fib({n}) slow = {fib_slow(n)}")
slow_time = time.time() - start

start = time.time()
print(f"fib({n}) fast = {fib_fast(n)}")
fast_time = time.time() - start

print(f"
Slow: {slow_time:.4f}s")
print(f"Fast: {fast_time:.6f}s")
print(f"Speedup: {slow_time/fast_time:.0f}x faster")
```

**Why is memoization so powerful?** `fib_slow(35)` calls itself about 29 million times. `fib_fast(35)` only calculates each value once and reuses it — so it calls itself exactly 35 times. Python's standard library has this built-in: `from functools import lru_cache`.
