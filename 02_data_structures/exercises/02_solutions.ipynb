### Exercise 1 — List manipulation

**Concept:** `append()` adds to the end; to remove by position use `.pop(index)`. To remove by value, use `.remove(value)`.

```python
movies = ["Inception", "The Matrix", "Interstellar", "Parasite", "Dune"]

# Append one more
movies.append("Oppenheimer")
print("After append:", movies)

# Remove the second one (index 1)
movies.pop(1)       # removes "The Matrix" at index 1
print("After removing 2nd:", movies)

# Alternative: del movies[1]  — same result
```

**Why `.pop(index)`?** `.pop()` removes by index and returns the removed item (useful if you need it). `.remove()` removes by value, which is handy when you know what to remove but not where it is.

### Exercise 2 — Sorted without modifying original

**Concept:** `.sort()` modifies the list IN-PLACE (changes the original). `sorted()` returns a NEW list and leaves the original untouched.

```python
nums = [5, 3, 8, 1, 9, 2, 7]

# sorted() — returns a new sorted list, original unchanged
descending = sorted(nums, reverse=True)

print("Original:", nums)         # unchanged
print("Descending:", descending) # new list

# Common mistake: nums.sort(reverse=True) WOULD work but modifies nums
# Test it:
original = [5, 3, 8, 1, 9, 2, 7]
original.sort(reverse=True)
print("After .sort():", original)   # original is now changed!
```

### Exercise 3 — Dictionary operations and iteration

**Concept:** `dict[key] = value` adds or updates a key. `.items()` returns all key-value pairs as tuples, perfect for looping.

```python
person = {
    "name": "Alice",
    "age": 28,
    "city": "London",
    "job": "Data Scientist"
}

# Add salary key
person["salary"] = 75000

# Iterate and print all key-value pairs
print("Person profile:")
for key, value in person.items():
    print(f"  {key}: {value}")

# Bonus: formatted nicely
print()
for key, value in person.items():
    print(f"  {key.capitalize():<10} -> {value}")
```

### Exercise 4 — Set intersection to find common elements

**Concept:** Converting lists to sets and using `&` (intersection) gives you elements that appear in BOTH. This is much cleaner than nested loops.

```python
list1 = [1, 2, 3, 4, 5, 6]
list2 = [4, 5, 6, 7, 8, 9]

# Method 1: Set intersection
common = set(list1) & set(list2)
print("Common elements:", common)   # {4, 5, 6}
print("As sorted list:", sorted(common))

# Method 2: List comprehension (also works, but slower for large lists)
common2 = [x for x in list1 if x in list2]
print("Method 2:", common2)

# Why sets are faster: set membership test is O(1), list is O(n)
# For small lists it doesn't matter; for millions of items, use sets.
```

### Exercise 5 — Word frequency counter

**Concept:** Iterate through words and build a dictionary. For each word: if it exists, increment by 1; if not, start at 1. The `.get(key, default)` method is perfect for this.

```python
sentence = 'the cat sat on the mat and the cat ate the rat'
words = sentence.split()   # split on spaces

# Method 1: Using dict.get() with default
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1   # get current count (0 if missing), add 1

print("Word frequencies:")
for word, count in sorted(freq.items(), key=lambda x: x[1], reverse=True):
    print(f"  '{word}': {count}")

print()
# Method 2: Using collections.Counter (professional way)
from collections import Counter
freq2 = Counter(words)
print("Most common:", freq2.most_common(5))
```

**`freq.get(word, 0) + 1` explained:** `dict.get(key, default)` returns the value if key exists, otherwise returns the default. So if 'the' is already in the dict with count 3, it returns 3. Then we add 1 to get 4, and store it back.

### Exercise 6 — Challenge: Find highest average grade

**Concept:** Loop through the list of dicts, compute the average for each student using `sum()/len()`, and track which student has the highest average.

```python
students = [
    {'name': 'Alice', 'grades': [85, 92, 88]},
    {'name': 'Bob',   'grades': [70, 75, 80]},
    {'name': 'Carol', 'grades': [95, 91, 97]},
]

# Method 1: Manual loop — easy to follow
best_student = None
best_avg = 0

for student in students:
    avg = sum(student['grades']) / len(student['grades'])
    print(f"{student['name']}: average = {avg:.1f}")
    if avg > best_avg:
        best_avg = avg
        best_student = student['name']

print(f"
Top student: {best_student} with average {best_avg:.1f}")

# Method 2: One-liner using max() with a key function
top = max(students, key=lambda s: sum(s['grades']) / len(s['grades']))
print(f"One-liner result: {top['name']}")
```

**`max()` with a key function:** `max()` can find the maximum of a complex object if you tell it *what to compare*. The `key=` argument takes a function that extracts the comparison value from each element.
