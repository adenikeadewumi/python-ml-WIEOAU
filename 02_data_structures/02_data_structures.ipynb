## 2.1 Lists

**What is a list?**
A list is an ordered, mutable (changeable) sequence of items. Items can be of any type, and you can mix types in one list. Lists are the most commonly used data structure in Python.

**When to use:** Whenever you have a collection of items whose order matters, or items you need to add/remove/change — student records, daily temperatures, items in a shopping cart.

**Key vocabulary:**
- *Ordered* means the items maintain the position you put them in
- *Mutable* means you can change, add, or remove items after creation
- *Index* is the position of an item — Python starts counting from 0

```python
# Creating lists
fruits = ["apple", "banana", "cherry"]
numbers = [3, 1, 4, 1, 5, 9, 2, 6]
mixed = [42, "hello", True, 3.14, None]   # different types in one list
empty = []

# Lists maintain order
print("Fruits:", fruits)
print("First fruit:", fruits[0])      # index 0 = first
print("Last fruit:", fruits[-1])      # index -1 = last

# Slicing works like strings
print("First two:", fruits[:2])       # ['apple', 'banana']
print("Last two:", fruits[-2:])       # ['banana', 'cherry']
```

```python
# Mutability — lists can be changed after creation
fruits = ["apple", "banana", "cherry"]
fruits[0] = "avocado"         # replace by index
fruits.append("mango")        # add to end
fruits.insert(1, "blueberry") # insert at position 1
fruits.remove("banana")       # remove first occurrence of this value
popped = fruits.pop()         # remove and return last item
print("Final list:", fruits)
print("Popped item:", popped)

# Useful list operations
numbers = [5, 3, 8, 1, 9, 2]
print("Length:", len(numbers))
print("Max:", max(numbers))
print("Min:", min(numbers))
print("Sum:", sum(numbers))
print("Sorted:", sorted(numbers))      # returns new sorted list
numbers.sort()                         # sorts in-place
print("After .sort():", numbers)
```

```python
# Checking membership and counting
fruits = ["apple", "banana", "cherry", "banana", "apple"]
print("apple" in fruits)         # True
print("grape" in fruits)         # False
print(fruits.count("banana"))    # 2 — how many times it appears
print(fruits.index("cherry"))    # 2 — index of first occurrence

# 2D lists (list of lists) — common in data science
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print("Row 1:", matrix[0])        # [1, 2, 3]
print("Element [1][2]:", matrix[1][2])   # 6 (row 1, column 2)
```

## 2.2 Tuples

**What is a tuple?**
A tuple is like a list but **immutable** — once created, it cannot be changed. This is its defining feature.

**When to use tuples:**
- When data should not change: GPS coordinates, RGB colours, database records
- As dictionary keys (lists cannot be dict keys because they are mutable)
- For function return values when returning multiple items
- They signal to other developers that this data is fixed

**Performance note:** Tuples are slightly faster than lists and use less memory. For fixed data, prefer tuples.

```python
# Creating tuples
point = (3, 7)
rgb_red = (255, 0, 0)
person = ("Alice", 25, "London")

# Single-element tuple MUST have a trailing comma
single = (42,)      # <- the comma makes it a tuple
not_tuple = (42)    # <- this is just 42 in parentheses (an int)
print(type(single), type(not_tuple))

# Indexing works the same as lists
print(point[0], point[1])

# Tuple unpacking — very Pythonic
x, y = point
r, g, b = rgb_red
name, age, city = person
print(f"x={x}, y={y}")
print(f"RGB: R={r} G={g} B={b}")
print(f"{name} is {age} and lives in {city}")
```

```python
# Immutability — trying to change a tuple raises an error
coords = (48.8566, 2.3522)   # Paris coordinates

# coords[0] = 51.5  # This would raise: TypeError: 'tuple' object does not support item assignment

# This is intentional — GPS coordinates should not be accidentally changed!

# Tuples as dictionary keys (lists cannot do this)
distances = {
    (0, 0): "Origin",
    (1, 0): "One step right",
    (0, 1): "One step up"
}
print(distances[(0, 0)])

# Swap using tuple unpacking
a, b = 10, 20
a, b = b, a   # Python creates a tuple (b, a) then unpacks it
print(f"Swapped: a={a}, b={b}")
```

## 2.3 Dictionaries

**What is a dictionary?**
A dictionary stores **key-value pairs**. Think of it like a real dictionary: you look up a *word* (key) to find its *definition* (value). Looking up a value by key is extremely fast — O(1) regardless of size.

**When to use:** Whenever you need to look things up by a label rather than by position. User profiles, configuration settings, counting occurrences, mapping one value to another.

**Rules:**
- Keys must be unique (duplicates overwrite the earlier value)
- Keys must be immutable (strings, numbers, tuples — not lists)
- Values can be anything, including other dicts or lists
- As of Python 3.7+, dictionaries maintain insertion order

```python
# Creating a dictionary
student = {
    "name": "Alice",
    "age": 22,
    "grades": [88, 92, 79],    # value can be a list
    "active": True
}

# Accessing values
print(student["name"])               # 'Alice'
# print(student["gpa"])              # KeyError if key doesn't exist!

# Safe access with .get() — returns None (or a default) if key missing
print(student.get("gpa"))            # None
print(student.get("gpa", "N/A"))     # 'N/A' — custom default

# Adding and updating entries
student["gpa"] = 3.8            # add new key
student["age"] = 23             # update existing key
print(student)
```

```python
# Iterating over a dictionary
student = {"name": "Alice", "age": 23, "city": "London"}

# Iterate over keys
for key in student:
    print(key)

# Iterate over values
for value in student.values():
    print(value)

# Iterate over key-value pairs (most common)
for key, value in student.items():
    print(f"  {key}: {value}")

# Check key existence
print("name" in student)    # True
print("gpa" in student)     # False

# Remove entries
del student["city"]
age = student.pop("age")    # remove and return value
print(student)
```

## 2.4 Sets

**What is a set?**
A set is an **unordered collection of unique items**. Adding a duplicate is silently ignored — the set just keeps one copy.

**When to use:**
- Remove duplicates from a list
- Fast membership testing (much faster than a list for large collections)
- Set operations: find items in common (intersection), combined items (union), etc.

**Key properties:**
- *Unordered* — you cannot rely on any particular order
- *Unique* — duplicates are automatically discarded
- *Mutable* — you can add/remove items
- *Fast* — membership test is O(1) (instant), vs O(n) for lists

```python
# Creating sets
primes = {2, 3, 5, 7, 11, 13}
evens  = {2, 4, 6, 8, 10, 12}

# Sets automatically eliminate duplicates
dirty_list = [1, 2, 2, 3, 3, 3, 4, 5, 5]
clean_set = set(dirty_list)
print("Unique values:", clean_set)    # order is not guaranteed!
print("Back to list:", sorted(clean_set))   # sort for display

# Fast membership testing
print(7 in primes)    # True — instant even with millions of items
print(4 in primes)    # False

# Adding and removing
primes.add(17)
primes.discard(2)    # remove if present (no error if missing)
print(primes)
```

```python
# Set operations — these are powerful for data analysis
primes = {2, 3, 5, 7, 11}
evens  = {2, 4, 6, 8, 10}

print("Intersection (&) — in BOTH sets:  ", primes & evens)   # {2}
print("Union       (|) — in EITHER set:  ", primes | evens)
print("Difference  (-) — in primes not evens:", primes - evens)
print("Sym diff    (^) — in one but not both:", primes ^ evens)

# Real use: find customers who bought both products
product_a_buyers = {"Alice", "Bob", "Carol", "David"}
product_b_buyers = {"Bob", "David", "Eve", "Frank"}

both = product_a_buyers & product_b_buyers
only_a = product_a_buyers - product_b_buyers
print("Bought both:", both)
print("Only bought A:", only_a)
```

## 2.5 Choosing the Right Data Structure

| Your situation | Best choice | Why |
|---------------|-------------|-----|
| Ordered, changeable collection | `list` | Append, insert, remove, index |
| Fixed data that shouldn't change | `tuple` | Immutable, faster, can be dict key |
| Look up values by a label | `dict` | O(1) lookup by key |
| Need unique values or fast membership | `set` | Auto-deduplication, O(1) `in` check |

---

## Key Takeaways

- `list` → ordered, mutable, allows duplicates → `[1, 2, 3]`
- `tuple` → ordered, immutable → `(1, 2, 3)`
- `dict` → key-value pairs, O(1) lookup → `{"key": "value"}`
- `set` → unique items, O(1) membership test → `{1, 2, 3}`
