### Exercise 1 — Diagonal matrix

**Concept:** `np.diag()` can both extract the diagonal of a matrix AND create a matrix from a 1D array. When given a 1D array, it places those values on the diagonal and fills the rest with zeros.

```python
import numpy as np

# Method 1: np.diag() — clean and idiomatic
matrix = np.diag([1, 2, 3, 4, 5])
print("Diagonal matrix:")
print(matrix)

# Method 2: Build it manually to understand what's happening
matrix2 = np.zeros((5, 5), dtype=int)
for i in range(5):
    matrix2[i, i] = i + 1
print("
Same matrix (built manually):")
print(matrix2)

# Verify they are equal
print("
Are they equal?", np.array_equal(matrix, matrix2))
```

### Exercise 2 — Standard deviation intervals

**Concept:** The empirical rule (68-95-99.7 rule) says that in a normal distribution, ~68% of values fall within 1 standard deviation of the mean. NumPy boolean indexing lets us count this precisely.

```python
import numpy as np
rng = np.random.default_rng(42)
data = rng.normal(loc=50, scale=10, size=1000)

mean = data.mean()
std = data.std()
n = len(data)

print(f"Mean: {mean:.2f}, Std: {std:.2f}")
print()

for num_std in [1, 2, 3]:
    # Boolean mask: True where |x - mean| <= num_std * std
    within = np.abs(data - mean) <= num_std * std
    count = within.sum()
    proportion = count / n * 100
    print(f"Within {num_std} std: {count}/{n} ({proportion:.1f}%)  [expected ~{[68,95,99.7][num_std-1]}%]")
```

**`np.abs(data - mean)`:** This is vectorised — it subtracts the mean from every element at once (broadcasting), then takes absolute value of every element. No loop needed. Result is an array of distances from the mean.

### Exercise 3 — Row-wise z-score normalisation

**Concept:** Z-score = (value - mean) / std. When we compute mean/std with `axis=1`, we get one value per row. Using `keepdims=True` preserves the shape so broadcasting works correctly.

```python
import numpy as np
rng = np.random.default_rng(42)
matrix = rng.integers(1, 100, size=(4, 5)).astype(float)

print("Original matrix:")
print(matrix)

# Compute row means and stds with keepdims=True
# keepdims=True: shape stays (4,1) instead of (4,) — needed for broadcasting
row_means = matrix.mean(axis=1, keepdims=True)   # shape: (4, 1)
row_stds  = matrix.std(axis=1, keepdims=True)    # shape: (4, 1)

normalised = (matrix - row_means) / row_stds

print("
Normalised (row z-scores):")
print(normalised.round(3))

# Verify: each row should have mean ~0 and std ~1
print("
Verification:")
print("Row means:", normalised.mean(axis=1).round(10))
print("Row stds: ", normalised.std(axis=1).round(3))
```

**Why `keepdims=True`?** Without it, `matrix.mean(axis=1)` returns shape `(4,)`. When you subtract shape `(4,5) - (4,)`, NumPy does not broadcast the way you expect. With `keepdims=True`, the result is shape `(4,1)`, and `(4,5) - (4,1)` broadcasts perfectly — subtracting each row's mean from every element in that row.

### Exercise 4 — Cosine similarity

**Concept:** Cosine similarity measures the angle between two vectors. It equals 1 for identical direction, 0 for perpendicular, -1 for opposite. It is used extensively in NLP (word vectors) and recommendation systems.

```python
import numpy as np

def cosine_similarity(u, v):
    """
    Compute cosine similarity between vectors u and v.
    
    Formula: cos(u, v) = (u . v) / (||u|| * ||v||)
    where . is dot product and ||x|| is the Euclidean norm (length).
    """
    dot_product = np.dot(u, v)           # u . v
    norm_u = np.linalg.norm(u)           # ||u||
    norm_v = np.linalg.norm(v)           # ||v||
    return dot_product / (norm_u * norm_v)

u = np.array([1, 2, 3, 4])
v = np.array([4, 3, 2, 1])

similarity = cosine_similarity(u, v)
print(f"u = {u}")
print(f"v = {v}")
print(f"Cosine similarity: {similarity:.4f}")

# Verify with identical vectors (should be 1.0)
print(f"Same vector: {cosine_similarity(u, u):.4f}")

# Perpendicular vectors (should be 0.0)
a = np.array([1, 0])
b = np.array([0, 1])
print(f"Perpendicular: {cosine_similarity(a, b):.4f}")
```
