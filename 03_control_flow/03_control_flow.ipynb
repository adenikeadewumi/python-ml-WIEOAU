## 3.1 Conditional Statements (if / elif / else)

**What is control flow?**
By default, Python runs your code line by line, top to bottom. Control flow statements let you change this — executing different blocks based on conditions, or repeating blocks multiple times.

**How `if/elif/else` works:**
1. Python evaluates the first `if` condition
2. If it is True, runs that block and skips the rest
3. If False, checks `elif` (else if) conditions in order
4. If none match, runs the `else` block (if provided)

**Important:** Only ONE branch ever runs — as soon as a True condition is found, the rest are skipped.

```python
score = 78

# The conditions are checked in order
if score >= 90:
    grade = "A"
elif score >= 80:       # only checked if first condition was False
    grade = "B"
elif score >= 70:       # only checked if both above were False
    grade = "C"
elif score >= 60:
    grade = "D"
else:                   # runs if ALL conditions above were False
    grade = "F"

print(f"Score: {score} -> Grade: {grade}")

# You can have just 'if' alone (no elif/else required)
if score > 100:
    print("Invalid score!")   # only prints if the condition is True
```

```python
# Nested conditions
username = "admin"
password = "correct_password"
is_logged_in = True

if is_logged_in:
    if username == "admin":
        print("Welcome, Administrator!")
    else:
        print(f"Welcome, {username}!")
else:
    print("Please log in first.")

# Ternary (one-line) if/else — good for simple assignments
x = 7
label = "even" if x % 2 == 0 else "odd"
print(f"{x} is {label}")

# Truthiness in conditions — you don't always need == True
name = "Alice"
if name:    # this checks if name is non-empty (truthy)
    print(f"Hello, {name}")

empty = ""
if not empty:
    print("Name is empty!")
```

## 3.2 For Loops

**What is a for loop?**
A for loop repeats a block of code once for each item in a sequence. The loop variable takes the value of each item in turn.

**You can loop over anything iterable:** lists, tuples, strings, dictionaries, `range()`, and many more.

**`range(start, stop, step)`** generates a sequence of numbers:
- `range(5)` → 0, 1, 2, 3, 4
- `range(1, 6)` → 1, 2, 3, 4, 5
- `range(0, 10, 2)` → 0, 2, 4, 6, 8

```python
# Basic for loop
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"  I like {fruit}")

# Looping over a string (strings are sequences of characters)
for char in "Python":
    print(char, end=" ")   # end=" " keeps on same line
print()   # newline

# range() — generate numbers to loop over
for i in range(5):            # 0, 1, 2, 3, 4
    print(i, end=" ")
print()

for i in range(1, 6):         # 1, 2, 3, 4, 5
    print(i, end=" ")
print()

for i in range(0, 11, 2):     # 0, 2, 4, 6, 8, 10
    print(i, end=" ")
print()
```

```python
# enumerate() — get index AND value simultaneously
# Without enumerate: fruits[i] requires knowing the index
# With enumerate: you get both at once, cleanly

fruits = ["apple", "banana", "cherry"]
for i, fruit in enumerate(fruits):
    print(f"  [{i}] {fruit}")

# You can set the starting number
for i, fruit in enumerate(fruits, start=1):
    print(f"  Fruit #{i}: {fruit}")
```

```python
# zip() — loop over two lists in parallel
# Like a zipper, pairs up items by position

names  = ["Alice", "Bob", "Carol"]
scores = [88, 72, 95]
grades = ["B", "C", "A"]

for name, score, grade in zip(names, scores, grades):
    print(f"  {name}: {score}/100 ({grade})")

# zip stops at the shortest list
longer  = [1, 2, 3, 4, 5]
shorter = ["a", "b", "c"]
print(list(zip(longer, shorter)))   # only 3 pairs
```

## 3.3 While Loops

**What is a while loop?**
A while loop repeats its body as long as its condition remains True. Unlike a for loop (which loops a fixed number of times over a sequence), a while loop continues until the condition becomes False.

**When to use while loops:**
- When you don't know in advance how many iterations you need
- Reading until end of file
- Game loops: "keep playing until player quits"
- Retrying an operation until it succeeds

**Critical warning:** Always make sure the condition will eventually become False. If it never does, your program runs forever — an "infinite loop".

```python
# Basic while loop
count = 0
while count < 5:
    print(f"  count = {count}")
    count += 1   # IMPORTANT: always update the variable in the condition!

print("Loop finished")

# Simulating a retry loop
attempts = 0
max_attempts = 3
success = False

while attempts < max_attempts and not success:
    attempts += 1
    print(f"  Attempt {attempts}...")
    if attempts == 2:     # simulate success on 2nd attempt
        success = True
        print("  Success!")

if not success:
    print("  All attempts failed.")
```

```python
# break — immediately exits the loop, no matter the condition
# continue — skips to the next iteration, ignores remaining code in loop body

print("break example:")
for n in range(10):
    if n == 5:
        print(f"  Found 5, stopping!")
        break            # exit loop entirely
    print(f"  n = {n}")

print()
print("continue example (skip even numbers):")
for n in range(10):
    if n % 2 == 0:
        continue         # skip back to top of loop
    print(f"  {n}", end=" ")   # only odd numbers reach this line
print()
```

## 3.4 Nested Loops and Patterns

**Nested loops** are loops inside loops. The inner loop completes ALL its iterations for each single iteration of the outer loop.

**Example:** If the outer loop runs 3 times and the inner runs 3 times, the body runs 3 × 3 = 9 times total.

**When to use:** Processing 2D data (tables, matrices, images), generating combinations, comparing every item to every other item.

```python
# Multiplication table using nested loops
print("Multiplication table (3x3):")
for i in range(1, 4):          # outer loop: rows 1, 2, 3
    for j in range(1, 4):      # inner loop: columns 1, 2, 3
        print(f"  {i}x{j}={i*j}", end="  ")
    print()    # newline after each row

print()

# Practical example: process student data
students = [
    {"name": "Alice", "grades": [85, 92, 78]},
    {"name": "Bob",   "grades": [70, 88, 95]},
]

for student in students:
    avg = sum(student["grades"]) / len(student["grades"])
    status = "Pass" if avg >= 75 else "Fail"
    print(f"{student['name']}: avg={avg:.1f} -> {status}")
```

## Key Takeaways

- `if/elif/else` branches — only ONE branch runs
- `for` loops iterate over any sequence; use `range()` for numbers
- `enumerate()` gives index + value; `zip()` pairs two sequences
- `while` loops run until condition is False — always ensure termination
- `break` exits the loop; `continue` skips to the next iteration
