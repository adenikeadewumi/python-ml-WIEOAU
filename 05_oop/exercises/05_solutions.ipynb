### Exercise 1 — Rectangle class

**Concept:** A class bundles data (attributes) and behaviour (methods) together. `__init__` runs automatically when you create an instance. `__str__` controls how the object looks when printed.

```python
class Rectangle:
    """Represents a 2D rectangle with width and height."""
    
    def __init__(self, width, height):
        # Store the dimensions as instance attributes
        self.width = width
        self.height = height
    
    def area(self):
        """Return the area (width x height)."""
        return self.width * self.height
    
    def perimeter(self):
        """Return the perimeter (2 x (width + height))."""
        return 2 * (self.width + self.height)
    
    def is_square(self):
        """Return True if width equals height."""
        return self.width == self.height
    
    def __str__(self):
        return f"Rectangle({self.width}x{self.height})"
    
    def __repr__(self):
        return f"Rectangle(width={self.width}, height={self.height})"

# Tests
r = Rectangle(4, 6)
print(r)                  # uses __str__
print(f"Area: {r.area()}")
print(f"Perimeter: {r.perimeter()}")
print(f"Is square: {r.is_square()}")

square = Rectangle(5, 5)
print(f"{square} — is square: {square.is_square()}")
```

**`__str__` vs `__repr__`:** `__str__` is for humans (used by `print()`). `__repr__` is for developers — it should be unambiguous and ideally show how to recreate the object. When only `__repr__` is defined, Python uses it for both.

### Exercise 2 — Student class

**Concept:** An instance attribute can be a mutable container like a list. Methods can modify that list or compute values from it.

```python
class Student:
    """Represents a student with a name and a list of grades."""
    
    def __init__(self, name):
        self.name = name
        self.grades = []   # starts empty
    
    def add_grade(self, grade):
        """Add a grade to the student's record."""
        if not 0 <= grade <= 100:
            raise ValueError(f"Grade must be 0-100, got {grade}")
        self.grades.append(grade)
    
    def average(self):
        """Return the average grade, or None if no grades."""
        if not self.grades:
            return None
        return sum(self.grades) / len(self.grades)
    
    def highest(self):
        return max(self.grades) if self.grades else None
    
    def lowest(self):
        return min(self.grades) if self.grades else None
    
    def __str__(self):
        avg = self.average()
        avg_str = f"{avg:.1f}" if avg else "N/A"
        return f"Student({self.name}, avg={avg_str}, grades={self.grades})"

# Test
alice = Student("Alice")
for g in [85, 92, 78, 95, 88]:
    alice.add_grade(g)

print(alice)
print(f"Average: {alice.average():.1f}")
print(f"Highest: {alice.highest()}")
print(f"Lowest:  {alice.lowest()}")

# Test edge case: empty student
bob = Student("Bob")
print(f"Bob average: {bob.average()}")
```

### Exercise 3 — Challenge: Stack class

**Concept:** A Stack is a Last-In-First-Out (LIFO) data structure. Think of a stack of plates — you can only add/remove from the top. Internally we use a list, but the interface restricts how it is used.

```python
class Stack:
    """
    A LIFO (Last-In, First-Out) stack.
    
    Think of it as a stack of plates: you can only add to the top
    and remove from the top.
    """
    
    def __init__(self):
        self._items = []   # _ prefix: internal detail, don't access directly
    
    def push(self, item):
        """Add an item to the top of the stack."""
        self._items.append(item)
    
    def pop(self):
        """Remove and return the top item. Raises IndexError if empty."""
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self._items.pop()   # list.pop() removes the last item
    
    def peek(self):
        """Return the top item WITHOUT removing it."""
        if self.is_empty():
            raise IndexError("peek at empty stack")
        return self._items[-1]   # -1 = last element
    
    def is_empty(self):
        """Return True if no items in stack."""
        return len(self._items) == 0
    
    def size(self):
        """Return number of items in stack."""
        return len(self._items)
    
    def __repr__(self):
        return f"Stack({self._items} <- top)"

# Tests
s = Stack()
print("Empty?", s.is_empty())

for item in [1, 2, 3, 4, 5]:
    s.push(item)

print(s)
print("Peek:", s.peek())      # 5
print("Pop:", s.pop())        # 5
print("Pop:", s.pop())        # 4
print("Size:", s.size())      # 3
print(s)

# Real use case: checking balanced brackets
def is_balanced(text):
    """Check if brackets in text are balanced using a stack."""
    stack = Stack()
    pairs = {')': '(', ']': '[', '}': '{'}
    
    for char in text:
        if char in '([{':
            stack.push(char)
        elif char in ')]}':
            if stack.is_empty() or stack.pop() != pairs[char]:
                return False
    return stack.is_empty()

tests = ["(1 + 2) * [3 + {4}]", "((())", "Hello World", "[(])"]
for t in tests:
    print(f"  '{t}' -> balanced: {is_balanced(t)}")
```

**Why encapsulate with a class?** We could just use a list directly. But a Stack class enforces the LIFO contract — users can only push/pop/peek. They cannot insert in the middle or reverse. This prevents bugs and makes code intentions clear.
