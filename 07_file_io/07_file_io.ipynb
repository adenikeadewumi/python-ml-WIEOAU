# 07 — File I/O & Working with Data Files
**Learning Objectives:** read/write text, CSV, JSON, pathlib
---

## 7.1 Reading and Writing Text Files

```python
# Write a file
with open("/tmp/demo.txt", "w") as f:
    f.write("Line 1\n")
    f.write("Line 2\n")
    f.writelines(["Line 3\n", "Line 4\n"])

# Read entire file
with open("/tmp/demo.txt") as f:
    content = f.read()
print(repr(content))

# Read line by line (memory efficient)
with open("/tmp/demo.txt") as f:
    for i, line in enumerate(f, 1):
        print(f"{i}: {line.strip()}")
```

## 7.2 Working with CSV Files

```python
import csv

# Write CSV
students = [
    ["Name", "Age", "Grade"],
    ["Alice", 22, "A"],
    ["Bob", 23, "B"],
    ["Carol", 21, "A+"],
]

with open("/tmp/students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(students)

# Read CSV as dicts (most useful)
with open("/tmp/students.csv") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(dict(row))
```

## 7.3 Working with JSON

```python
import json

# Write JSON
config = {
    "model": "random_forest",
    "hyperparameters": {"n_estimators": 100, "max_depth": 5},
    "features": ["age", "income", "score"],
    "version": 1.2
}

with open("/tmp/config.json", "w") as f:
    json.dump(config, f, indent=2)

# Read JSON
with open("/tmp/config.json") as f:
    loaded = json.load(f)

print(loaded["model"])
print(loaded["hyperparameters"])
```

## 7.4 pathlib — Modern Path Handling

```python
from pathlib import Path

# Creating paths
p = Path("/tmp")
config_path = p / "config.json"    # OS-independent path joining
print(config_path)
print(config_path.exists())
print(config_path.suffix)          # .json
print(config_path.stem)            # config
print(config_path.parent)          # /tmp

# List files matching pattern
tmp = Path("/tmp")
for f in tmp.glob("*.txt"):
    print(f.name, f.stat().st_size, "bytes")
```

---
## 🧩 Key Takeaways
- Always use `with open(...)` — it closes files automatically
- `csv.DictReader` reads CSV rows as dictionaries
- `json.dump()/load()` for serialising Python data structures
- `pathlib.Path` is the modern, OS-independent way to work with paths
